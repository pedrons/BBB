```python
# Dependência para receber eventos WebSocket do ComfyUI.
# Se necessário, execute antes em outra célula:
# %pip install -q websocket-client

import io
import json
import os
import shutil
import time
import uuid
import threading
from pathlib import Path

import gradio as gr
import requests
import websocket

v2v_frames = gr.Dropdown(
    choices=[81],
    value=81,
    label="Quadros por trecho — 6 trechos sequenciais",
)

v2v_button = gr.Button(
    "Gerar vídeo de 30 segundos",
    variant="primary",
)

from PIL import Image, ImageOps


# Ajuste estes valores se o ComfyUI estiver em outro endereço ou pasta.
COMFYUI_URL = os.environ.get("COMFYUI_URL", "http://127.0.0.1:8188").rstrip("/")
COMFYUI_DIR = Path(os.environ.get("COMFYUI_DIR", "ComfyUI_repo"))

OUTPUT_DIR = Path("outputs/gradio")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# Os arquivos precisam existir nas pastas de modelos do ComfyUI.
VACE_MODEL = "wan2.1_vace_14B_fp16.safetensors"
WAN_TEXT_ENCODER = "umt5_xxl_fp8_e4m3fn_scaled.safetensors"
WAN_VAE = "wan_2.1_vae.safetensors"

FLUX_FILL_MODEL = "flux1-fill-dev.safetensors"
FLUX_CLIP_L = "clip_l.safetensors"
FLUX_T5 = "t5xxl_fp8_e4m3fn.safetensors"
FLUX_VAE = "ae.safetensors"

NEGATIVE_PROMPT_VACE = (
    "flicker, jitter, blurry, low quality, distorted, deformed anatomy, "
    "extra limbs, duplicate objects, text, subtitles, watermark"
)

# Não deixa duas gerações grandes competirem pela VRAM ao mesmo tempo.
GPU_LOCK = threading.Lock()


def check_comfyui():
    try:
        response = requests.get(f"{COMFYUI_URL}/system_stats", timeout=10)
        response.raise_for_status()
    except requests.RequestException as exc:
        raise gr.Error(
            f"Não consegui acessar o ComfyUI em {COMFYUI_URL}. "
            "Confira se o servidor está ativo."
        ) from exc


def upload_image_to_comfyui(image: Image.Image) -> str:
    buffer = io.BytesIO()
    image.save(buffer, format="PNG")
    buffer.seek(0)

    filename = f"gradio_{uuid.uuid4().hex[:12]}.png"

    try:
        response = requests.post(
            f"{COMFYUI_URL}/upload/image",
            files={"image": (filename, buffer.getvalue(), "image/png")},
            data={"type": "input", "overwrite": "true"},
            timeout=120,
        )
        response.raise_for_status()
    except requests.RequestException as exc:
        raise gr.Error(f"Falha ao enviar imagem para o ComfyUI: {exc}") from exc

    return response.json().get("name", filename)


def copy_video_to_comfyui_input(video_path) -> str:
    if isinstance(video_path, dict):
        video_path = video_path.get("path") or video_path.get("name")

    if not video_path or not Path(video_path).is_file():
        raise gr.Error("Não consegui ler o vídeo enviado.")

    input_dir = COMFYUI_DIR / "input"
    input_dir.mkdir(parents=True, exist_ok=True)

    source = Path(video_path)
    safe_name = f"gradio_{uuid.uuid4().hex[:12]}{source.suffix.lower()}"
    destination = input_dir / safe_name
    shutil.copy2(source, destination)

    return safe_name


def make_reference_board(face_ref, body_ref, location_ref):
    """Combina até três referências numa única imagem para o nó VACE."""
    references = [face_ref, body_ref, location_ref]
    references = [img.convert("RGB") for img in references if img is not None]

    if not references:
        return None

    tile_size = 512
    board = Image.new("RGB", (tile_size * 2, tile_size * 2), (128, 128, 128))
    positions = [(0, 0), (tile_size, 0), (0, tile_size)]

    for image, (x, y) in zip(references, positions):
        fitted = ImageOps.contain(image, (tile_size, tile_size))
        px = x + (tile_size - fitted.width) // 2
        py = y + (tile_size - fitted.height) // 2
        board.paste(fitted, (px, py))

    return board


def make_rgba_inpaint_image(image: Image.Image, mask_image: Image.Image) -> Image.Image:
    """
    Interpreta branco na máscara como área a editar e preto como área a preservar.
    Converte a máscara em canal alfa para que o LoadImage do ComfyUI gere a máscara.
    """
    rgb = image.convert("RGB")
    mask = mask_image.convert("L").resize(rgb.size, Image.Resampling.NEAREST)

    # No LoadImage do ComfyUI, o canal alfa é invertido para formar a máscara.
    alpha = ImageOps.invert(mask)
    rgba = rgb.convert("RGBA")
    rgba.putalpha(alpha)
    return rgba


def get_saved_media(node_output, media_type):
    """Localiza o arquivo que um nó SaveVideo ou SaveImage gravou."""
    for container in (node_output, node_output.get("ui", {})):
        if not isinstance(container, dict):
            continue

        items = container.get(media_type, [])
        if isinstance(items, dict):
            items = [items]

        for item in items:
            if isinstance(item, dict) and item.get("filename"):
                return item

    return None


def get_history_entry(prompt_id):
    response = requests.get(
        f"{COMFYUI_URL}/history/{prompt_id}",
        timeout=30,
    )
    response.raise_for_status()
    return response.json().get(prompt_id)


def extract_execution_error(entry):
    status = entry.get("status", {})
    for message in status.get("messages", []):
        if (
            isinstance(message, (list, tuple))
            and len(message) > 1
            and message[0] == "execution_error"
        ):
            detail = message[1]
            if isinstance(detail, dict):
                return (
                    detail.get("exception_message")
                    or detail.get("message")
                    or json.dumps(detail, ensure_ascii=False)
                )
            return str(detail)
    return None


def run_comfy_workflow(
    workflow,
    save_node_id,
    media_type,
    output_folder,
    node_labels,
    progress,
):
    """
    Envia o workflow e monitora a API WebSocket do ComfyUI.
    Retorna o caminho local do arquivo salvo.
    """
    check_comfyui()

    client_id = str(uuid.uuid4())
    ws_url = (
        COMFYUI_URL.replace("https://", "wss://")
        .replace("http://", "ws://")
        + f"/ws?clientId={client_id}"
    )

    try:
        ws = websocket.create_connection(ws_url, timeout=2)
    except Exception as exc:
        raise gr.Error(
            "Não consegui abrir o WebSocket do ComfyUI. "
            "Confira a URL e a porta do servidor."
        ) from exc

    prompt_id = None
    saved_file = None
    last_history_check = 0
    deadline = time.time() + 3 * 60 * 60

    try:
        response = requests.post(
            f"{COMFYUI_URL}/prompt",
            json={"prompt": workflow, "client_id": client_id},
            timeout=60,
        )

        if not response.ok:
            raise gr.Error(
                "O ComfyUI rejeitou o workflow:\n" + response.text[:5000]
            )

        submitted = response.json()
        if submitted.get("error"):
            details = submitted.get("node_errors", submitted["error"])
            raise gr.Error(
                "O ComfyUI rejeitou o workflow:\n"
                + json.dumps(details, ensure_ascii=False, indent=2)
            )

        prompt_id = submitted.get("prompt_id")
        if not prompt_id:
            raise gr.Error(f"Resposta inesperada do ComfyUI: {submitted}")

        progress(0.08, desc="Workflow enviado ao ComfyUI.")
        print(f"[ComfyUI] Workflow enviado: {prompt_id}", flush=True)

        while time.time() < deadline:
            try:
                raw_message = ws.recv()
            except websocket.WebSocketTimeoutException:
                raw_message = None
            except Exception:
                raw_message = None

            if isinstance(raw_message, str):
                try:
                    message = json.loads(raw_message)
                except json.JSONDecodeError:
                    message = {}

                event_type = message.get("type")
                data = message.get("data", {})

                if data.get("prompt_id") == prompt_id:
                    if event_type == "executing":
                        node_id = str(data.get("node"))
                        label = node_labels.get(node_id, f"Nó {node_id}")

                        if data.get("node") is None:
                            progress(0.94, desc="Execução terminada; recuperando o arquivo...")
                            print("[ComfyUI] Nós concluídos.", flush=True)
                        else:
                            progress(0.20, desc=f"Em execução: {label}")
                            print(f"[ComfyUI] Em execução: {label}", flush=True)

                    elif event_type == "progress":
                        value = int(data.get("value", 0))
                        maximum = max(1, int(data.get("max", 1)))
                        percent = min(1.0, value / maximum)
                        overall = 0.25 + 0.62 * percent
                        progress(
                            overall,
                            desc=f"Amostragem: passo {value}/{maximum}",
                        )
                        print(
                            f"[ComfyUI] Sampler: {value}/{maximum}",
                            flush=True,
                        )

                    elif event_type == "execution_error":
                        raise gr.Error(
                            "O ComfyUI encontrou um erro durante a execução:\n"
                            + json.dumps(data, ensure_ascii=False, indent=2)
                        )

            # Consulta também o histórico como fallback para versões sem eventos.
            if time.time() - last_history_check >= 3:
                last_history_check = time.time()
                entry = get_history_entry(prompt_id)

                if entry:
                    error = extract_execution_error(entry)
                    if error:
                        raise gr.Error(f"Erro no ComfyUI:\n{error}")

                    output = entry.get("outputs", {}).get(
                        str(save_node_id), {}
                    )
                    saved_file = get_saved_media(output, media_type)

                    if saved_file:
                        break

            time.sleep(0.2)

        if not saved_file:
            raise gr.Error(
                "A execução não retornou um arquivo antes do limite de espera. "
                "Confira a fila e o histórico do ComfyUI."
            )

    finally:
        try:
            ws.close()
        except Exception:
            pass

    progress(0.96, desc="Baixando o resultado do ComfyUI...")

    try:
        result = requests.get(
            f"{COMFYUI_URL}/view",
            params={
                "filename": saved_file["filename"],
                "subfolder": saved_file.get("subfolder", ""),
                "type": saved_file.get("type", "output"),
            },
            timeout=300,
        )
        result.raise_for_status()
    except requests.RequestException as exc:
        raise gr.Error(f"Falha ao baixar o resultado: {exc}") from exc

    extension = Path(saved_file["filename"]).suffix
    if not extension:
        extension = ".mp4" if media_type in ("videos", "gifs") else ".png"

    output_folder = Path(output_folder)
    output_folder.mkdir(parents=True, exist_ok=True)
    local_path = output_folder / f"{prompt_id}{extension}"
    local_path.write_bytes(result.content)

    progress(1.0, desc="Concluído.")
    print(f"[ComfyUI] Arquivo salvo em: {local_path}", flush=True)
    return str(local_path), prompt_id


def build_vace_workflow(
    video_filename,
    reference_filename,
    prompt,
    width,
    height,
    frames,
    steps,
    cfg,
    seed,
):
    return {
        "1": {
            "class_type": "UNETLoader",
            "inputs": {
                "unet_name": VACE_MODEL,
                "weight_dtype": "default",
            },
        },
        "2": {
            "class_type": "CLIPLoader",
            "inputs": {
                "clip_name": WAN_TEXT_ENCODER,
                "type": "wan",
                "device": "default",
            },
        },
        "3": {
            "class_type": "VAELoader",
            "inputs": {"vae_name": WAN_VAE},
        },
        "4": {
            "class_type": "LoadVideo",
            "inputs": {"file": video_filename},
        },
        "5": {
            "class_type": "GetVideoComponents",
            "inputs": {"video": ["4", 0]},
        },
        "6": {
            "class_type": "CLIPTextEncode",
            "inputs": {
                "clip": ["2", 0],
                "text": prompt,
            },
        },
        "7": {
            "class_type": "CLIPTextEncode",
            "inputs": {
                "clip": ["2", 0],
                "text": NEGATIVE_PROMPT_VACE,
            },
        },
        "8": {
            "class_type": "WanVaceToVideo",
            "inputs": {
                "positive": ["6", 0],
                "negative": ["7", 0],
                "vae": ["3", 0],
                "control_video": ["5", 0],
                "reference_image": (
                    ["9", 0] if reference_filename else None
                ),
                "width": int(width),
                "height": int(height),
                "length": int(frames),
                "batch_size": 1,
                "strength": 1.0,
            },
        },
        "9": {
            "class_type": "LoadImage",
            "inputs": {"image": reference_filename or ""},
        },
        "10": {
            "class_type": "ModelSamplingSD3",
            "inputs": {
                "model": ["1", 0],
                "shift": 8,
            },
        },
        "11": {
            "class_type": "KSampler",
            "inputs": {
                "model": ["10", 0],
                "positive": ["8", 0],
                "negative": ["8", 1],
                "latent_image": ["8", 2],
                "seed": int(seed),
                "steps": int(steps),
                "cfg": float(cfg),
                "sampler_name": "uni_pc",
                "scheduler": "simple",
                "denoise": 1.0,
            },
        },
        "12": {
            "class_type": "TrimVideoLatent",
            "inputs": {
                "samples": ["11", 0],
                "trim_amount": ["8", 3],
            },
        },
        "13": {
            "class_type": "VAEDecode",
            "inputs": {
                "samples": ["12", 0],
                "vae": ["3", 0],
            },
        },
        "14": {
            "class_type": "CreateVideo",
            "inputs": {
                "images": ["13", 0],
                "fps": ["5", 2],
                "codec": "none",
            },
        },
        "15": {
            "class_type": "SaveVideo",
            "inputs": {
                "video": ["14", 0],
                "filename_prefix": "gradio/vace_v2v",
                "format": "auto",
                "codec": "auto",
            },
        },
    }


def run_vace(
    video_path,
    face_ref,
    body_ref,
    location_ref,
    prompt,
    width,
    height,
    frames,
    steps,
    cfg,
    seed,
    progress=gr.Progress(),
):
    if not video_path:
        raise gr.Error("Envie o vídeo de origem.")
    if not prompt or not prompt.strip():
        raise gr.Error("Escreva um prompt para orientar a transformação.")

    width = max(16, (int(width) // 16) * 16)
    height = max(16, (int(height) // 16) * 16)
    frames = max(5, 4 * ((int(frames) - 1) // 4) + 1)

    if seed is None or int(seed) < 0:
        seed = int.from_bytes(os.urandom(8), "big") % (2**63 - 1)

    with GPU_LOCK:
        progress(0.02, desc="Preparando o vídeo e as referências...")
        video_filename = copy_video_to_comfyui_input(video_path)

        reference_board = make_reference_board(
            face_ref,
            body_ref,
            location_ref,
        )

        reference_filename = None
        if reference_board is not None:
            reference_filename = upload_image_to_comfyui(reference_board)

        workflow = build_vace_workflow(
            video_filename=video_filename,
            reference_filename=reference_filename,
            prompt=prompt.strip(),
            width=width,
            height=height,
            frames=frames,
            steps=int(steps),
            cfg=float(cfg),
            seed=int(seed),
        )

        node_labels = {
            "1": "carregando o modelo Wan VACE",
            "2": "carregando o encoder de texto",
            "3": "carregando o VAE",
            "4": "lendo o vídeo",
            "5": "extraindo quadros do vídeo",
            "6": "codificando o prompt",
            "8": "preparando condicionamento VACE",
            "11": "gerando os quadros",
            "13": "decodificando os quadros",
            "14": "montando o vídeo",
            "15": "salvando o vídeo",
        }

        path, prompt_id = run_comfy_workflow(
            workflow=workflow,
            save_node_id="15",
            media_type="videos",
            output_folder=OUTPUT_DIR / "vace",
            node_labels=node_labels,
            progress=progress,
        )

    return path, f"VACE concluído. ID: {prompt_id}"


def build_flux_fill_workflow(image_filename, prompt, steps, seed):
    return {
        "1": {
            "class_type": "UNETLoader",
            "inputs": {
                "unet_name": FLUX_FILL_MODEL,
                "weight_dtype": "default",
            },
        },
        "2": {
            "class_type": "DualCLIPLoader",
            "inputs": {
                "clip_name1": FLUX_CLIP_L,
                "clip_name2": FLUX_T5,
                "type": "flux",
                "device": "default",
            },
        },
        "3": {
            "class_type": "VAELoader",
            "inputs": {"vae_name": FLUX_VAE},
        },
        "4": {
            "class_type": "LoadImage",
            "inputs": {"image": image_filename},
        },
        "5": {
            "class_type": "CLIPTextEncode",
            "inputs": {
                "clip": ["2", 0],
                "text": prompt,
            },
        },
        "6": {
            "class_type": "FluxGuidance",
            "inputs": {
                "conditioning": ["5", 0],
                "guidance": 30.0,
            },
        },
        "7": {
            "class_type": "ConditioningZeroOut",
            "inputs": {"conditioning": ["5", 0]},
        },
        "8": {
            "class_type": "DifferentialDiffusion",
            "inputs": {"model": ["1", 0]},
        },
        "9": {
            "class_type": "InpaintModelConditioning",
            "inputs": {
                "positive": ["6", 0],
                "negative": ["7", 0],
                "vae": ["3", 0],
                "pixels": ["4", 0],
                "mask": ["4", 1],
                "noise_mask": True,
            },
        },
        "10": {
            "class_type": "KSampler",
            "inputs": {
                "model": ["8", 0],
                "positive": ["9", 0],
                "negative": ["9", 1],
                "latent_image": ["9", 2],
                "seed": int(seed),
                "steps": int(steps),
                "cfg": 1.0,
                "sampler_name": "euler",
                "scheduler": "simple",
                "denoise": 1.0,
            },
        },
        "11": {
            "class_type": "VAEDecode",
            "inputs": {
                "samples": ["10", 0],
                "vae": ["3", 0],
            },
        },
        "12": {
            "class_type": "SaveImage",
            "inputs": {
                "images": ["11", 0],
                "filename_prefix": "gradio/flux_fill",
            },
        },
    }


def run_flux_fill(
    image,
    mask,
    prompt,
    steps,
    seed,
    progress=gr.Progress(),
):
    if image is None:
        raise gr.Error("Envie a imagem que deseja editar.")
    if mask is None:
        raise gr.Error("Envie uma máscara em preto e branco.")
    if not prompt or not prompt.strip():
        raise gr.Error("Escreva o que deve aparecer na área mascarada.")

    if seed is None or int(seed) < 0:
        seed = int.from_bytes(os.urandom(8), "big") % (2**63 - 1)

    with GPU_LOCK:
        progress(0.02, desc="Preparando imagem e máscara...")
        rgba_image = make_rgba_inpaint_image(image, mask)
        image_filename = upload_image_to_comfyui(rgba_image)

        workflow = build_flux_fill_workflow(
            image_filename=image_filename,
            prompt=prompt.strip(),
            steps=int(steps),
            seed=int(seed),
        )

        node_labels = {
            "1": "carregando Flux Fill",
            "2": "carregando encoders de texto",
            "3": "carregando VAE",
            "4": "lendo imagem e máscara",
            "5": "codificando o prompt",
            "9": "preparando inpainting",
            "10": "gerando a imagem",
            "11": "decodificando imagem",
            "12": "salvando resultado",
        }

        path, prompt_id = run_comfy_workflow(
            workflow=workflow,
            save_node_id="12",
            media_type="images",
            output_folder=OUTPUT_DIR / "flux_fill",
            node_labels=node_labels,
            progress=progress,
        )

    return path, path, f"Flux Fill concluído. ID: {prompt_id}"


with gr.Blocks(
    theme=gr.themes.Soft(),
    title="Wan VACE e Flux Fill",
) as demo:
    gr.Markdown(
        """
        # Painel de vídeo e edição de imagem

        O V2V usa Wan VACE para transformar um vídeo com orientação textual e
        referências visuais. O Flux Fill edita uma imagem usando uma máscara.
        O progresso do ComfyUI é exibido durante a execução.
        """
    )

    with gr.Tabs():
        with gr.Tab("Vídeo para vídeo — Wan/VACE"):
            gr.Markdown(
                "O VACE recebe um vídeo de controle e até três referências. "
                "As referências são reunidas numa montagem visual única."
            )

            with gr.Row():
                with gr.Column():
                    v2v_video = gr.Video(
                        label="Vídeo de origem",
                        format="mp4",
                        sources=["upload"],
                    )

                    v2v_prompt = gr.Textbox(
                        label="Prompt para a transformação",
                        placeholder=(
                            "Descreva a aparência desejada e o que deve mudar. "
                            "Ex.: manter o movimento da câmera e substituir "
                            "o cenário por uma floresta ao entardecer."
                        ),
                        lines=4,
                    )

                    with gr.Row():
                        v2v_face = gr.Image(
                            label="Referência de rosto (opcional)",
                            type="pil",
                        )
                        v2v_body = gr.Image(
                            label="Referência de corpo/roupa (opcional)",
                            type="pil",
                        )
                        v2v_location = gr.Image(
                            label="Referência de local/estilo (opcional)",
                            type="pil",
                        )

                    with gr.Row():
                        v2v_width = gr.Slider(
                            512, 1280, value=832, step=16,
                            label="Largura de saída",
                        )
                        v2v_height = gr.Slider(
                            384, 720, value=480, step=16,
                            label="Altura de saída",
                        )

                    v2v_frames = gr.Dropdown(
                        choices=[33, 49, 81],
                        value=49,
                        label="Frames (4n+1)",
                    )

                    with gr.Row():
                        v2v_steps = gr.Slider(
                            10, 40, value=20, step=1,
                            label="Passos",
                        )
                        v2v_cfg = gr.Slider(
                            1, 10, value=6, step=0.5,
                            label="CFG",
                        )

                    v2v_seed = gr.Number(
                        value=-1,
                        precision=0,
                        label="Seed (-1 para aleatória)",
                    )
                    v2v_button = gr.Button(
                        "Gerar vídeo V2V",
                        variant="primary",
                    )

                with gr.Column():
                    v2v_output = gr.Video(label="Vídeo gerado")
                    v2v_status = gr.Textbox(label="Status", lines=2)

            v2v_button.click(
                fn=run_vace,
                inputs=[
                    v2v_video,
                    v2v_face,
                    v2v_body,
                    v2v_location,
                    v2v_prompt,
                    v2v_width,
                    v2v_height,
                    v2v_frames,
                    v2v_steps,
                    v2v_cfg,
                    v2v_seed,
                ],
                outputs=[v2v_output, v2v_status],
            )

        with gr.Tab("Inpainting — Flux Fill"):
            gr.Markdown(
                "Envie uma imagem e uma máscara. **Branco = área a editar; "
                "preto = preservar.** A imagem resultante pode ser baixada e "
                "enviada depois como referência na aba V2V."
            )

            with gr.Row():
                with gr.Column():
                    flux_image = gr.Image(
                        label="Imagem original",
                        type="pil",
                    )
                    flux_mask = gr.Image(
                        label="Máscara em preto e branco",
                        type="pil",
                    )
                    flux_prompt = gr.Textbox(
                        label="O que deve aparecer na área branca?",
                        placeholder=(
                            "Ex.: preencher a área com uma jaqueta vermelha "
                            "de couro, mantendo iluminação e perspectiva."
                        ),
                        lines=3,
                    )
                    flux_steps = gr.Slider(
                        10, 40, value=20, step=1,
                        label="Passos",
                    )
                    flux_seed = gr.Number(
                        value=-1,
                        precision=0,
                        label="Seed (-1 para aleatória)",
                    )
                    flux_button = gr.Button(
                        "Editar imagem com Flux Fill",
                        variant="primary",
                    )

                with gr.Column():
                    flux_output = gr.Image(label="Imagem editada")
                    flux_download = gr.File(label="Baixar imagem editada")
                    flux_status = gr.Textbox(label="Status", lines=2)

            flux_button.click(
                fn=run_flux_fill,
                inputs=[
                    flux_image,
                    flux_mask,
                    flux_prompt,
                    flux_steps,
                    flux_seed,
                ],
                outputs=[
                    flux_output,
                    flux_download,
                    flux_status,
                ],
            )


# Em Colab, share=True cria um link público temporário.
# Use referências/vídeos não sensíveis ou troque para share=False
# se estiver acessando a interface por um túnel privado.
demo.queue().launch(share=True, debug=False)
```

```python
# Adicione junto aos imports da célula:
import subprocess
import cv2


FPS_FINAL = 16
FRAMES_POR_TRECHO = 81       # formato Wan: 4n + 1
NUM_TRECHOS = 6
DURACAO_FINAL_SEGUNDOS = 30


def executar_ffmpeg(args):
    """Executa ffmpeg e mostra uma mensagem compreensível em caso de falha."""
    if shutil.which("ffmpeg") is None:
        raise gr.Error("ffmpeg não foi encontrado no ambiente.")

    resultado = subprocess.run(
        ["ffmpeg", "-hide_banner", "-loglevel", "error", "-y", *args],
        capture_output=True,
        text=True,
    )

    if resultado.returncode != 0:
        raise gr.Error(
            "Falha no processamento de vídeo com ffmpeg:\n"
            + (resultado.stderr[-4000:] or "Sem detalhes adicionais.")
        )


def padronizar_video_controle(
    entrada,
    saida,
    largura,
    altura,
    fps=FPS_FINAL,
    quadros=FRAMES_POR_TRECHO,
):
    """
    Converte o vídeo de entrada para 16 fps e dimensões compatíveis.
    Se o vídeo for curto, repete o último quadro até completar o trecho.
    """
    filtro = (
        f"fps={fps},"
        f"scale={largura}:{altura}:force_original_aspect_ratio=decrease,"
        f"pad={largura}:{altura}:(ow-iw)/2:(oh-ih)/2,"
        f"tpad=stop_mode=clone:stop_duration=10,"
        f"trim=end_frame={quadros},"
        "setpts=PTS-STARTPTS,format=yuv420p"
    )

    executar_ffmpeg([
        "-i", str(entrada),
        "-vf", filtro,
        "-an",
        "-frames:v", str(quadros),
        "-c:v", "libx264",
        "-preset", "veryfast",
        "-pix_fmt", "yuv420p",
        str(saida),
    ])

    return str(saida)


def criar_controle_do_ultimo_quadro(
    imagem,
    saida,
    largura,
    altura,
    fps=FPS_FINAL,
    quadros=FRAMES_POR_TRECHO,
):
    """Cria um vídeo de controle estático a partir do último quadro anterior."""
    filtro = (
        f"scale={largura}:{altura}:force_original_aspect_ratio=decrease,"
        f"pad={largura}:{altura}:(ow-iw)/2:(oh-ih)/2,"
        f"fps={fps},"
        f"trim=end_frame={quadros},"
        "setpts=PTS-STARTPTS,format=yuv420p"
    )

    executar_ffmpeg([
        "-loop", "1",
        "-i", str(imagem),
        "-vf", filtro,
        "-frames:v", str(quadros),
        "-an",
        "-c:v", "libx264",
        "-preset", "veryfast",
        "-pix_fmt", "yuv420p",
        str(saida),
    ])

    return str(saida)


def extrair_ultimo_quadro(video_path, destino):
    captura = cv2.VideoCapture(str(video_path))
    if not captura.isOpened():
        raise gr.Error(f"Não consegui abrir o vídeo gerado: {video_path}")

    total = int(captura.get(cv2.CAP_PROP_FRAME_COUNT))
    if total <= 0:
        captura.release()
        raise gr.Error("O vídeo gerado não contém quadros legíveis.")

    captura.set(cv2.CAP_PROP_POS_FRAMES, total - 1)
    sucesso, quadro = captura.read()
    captura.release()

    if not sucesso:
        raise gr.Error("Não consegui extrair o último quadro do trecho.")

    quadro_rgb = cv2.cvtColor(quadro, cv2.COLOR_BGR2RGB)
    Image.fromarray(quadro_rgb).save(destino)
    return str(destino)


def juntar_trechos_em_30_segundos(videos, destino, fps=FPS_FINAL):
    """Concatena os trechos e limita o arquivo final a exatamente 30 segundos."""
    lista = Path(destino).with_suffix(".concat.txt")

    # O concat demuxer do ffmpeg usa aspas simples para os caminhos.
    linhas = []
    for caminho in videos:
        caminho_escapado = str(Path(caminho).resolve()).replace("'", "'\\''")
        linhas.append(f"file '{caminho_escapado}'")

    lista.write_text("\n".join(linhas) + "\n", encoding="utf-8")

    executar_ffmpeg([
        "-f", "concat",
        "-safe", "0",
        "-i", str(lista),
        "-vf", f"fps={fps}",
        "-t", str(DURACAO_FINAL_SEGUNDOS),
        "-an",
        "-c:v", "libx264",
        "-preset", "fast",
        "-crf", "18",
        "-pix_fmt", "yuv420p",
        "-movflags", "+faststart",
        str(destino),
    ])

    lista.unlink(missing_ok=True)
    return str(destino)


def prompt_do_trecho(prompt_base, indice):
    """Instrui a progressão da câmera entre os seis trechos."""
    etapas = [
        "Comece exatamente pelo enquadramento da perna do vídeo de controle. "
        "Inicie uma subida lenta e suave da câmera, aproximando-se do joelho.",

        "Continue do quadro de controle, sem corte. Suba lentamente da perna "
        "em direção à coxa, mantendo a mesma mulher, ambiente e iluminação.",

        "Continue o movimento da câmera para cima, revelando gradualmente "
        "a parte superior do corpo, com anatomia natural e continuidade visual.",

        "Continue a subida suave em direção aos ombros e ao pescoço. "
        "Não faça cortes, saltos de câmera ou mudanças bruscas de cenário.",

        "Continue o movimento e comece a revelar o rosto. Use a foto enviada "
        "somente como referência para a aparência facial.",

        "Termine a subida revelando claramente o rosto da mulher. "
        "Encerre em um close-up estável e natural, mantendo os traços faciais "
        "coerentes com a foto de referência.",
    ]

    instrucoes_gerais = (
        "Gere uma continuação visual do trecho anterior. A câmera deve avançar "
        "gradualmente; não repita o trecho anterior nem volte ao enquadramento "
        "inicial. Mantenha perspectiva, iluminação e aparência coerentes. "
        "Use a referência facial apenas para orientar o rosto."
    )

    return (
        f"{prompt_base.strip()}\n\n"
        f"{instrucoes_gerais}\n"
        f"Etapa {indice + 1} de {NUM_TRECHOS}: {etapas[indice]}"
    )


def build_vace_workflow(
    video_filename,
    reference_filename,
    prompt,
    width,
    height,
    frames,
    steps,
    cfg,
    seed,
):
    """
    Workflow API VACE. Se não houver referência, omite o input opcional
    reference_image em vez de enviar null.
    """
    workflow = {
        "1": {
            "class_type": "UNETLoader",
            "inputs": {
                "unet_name": VACE_MODEL,
                "weight_dtype": "default",
            },
        },
        "2": {
            "class_type": "CLIPLoader",
            "inputs": {
                "clip_name": WAN_TEXT_ENCODER,
                "type": "wan",
                "device": "default",
            },
        },
        "3": {
            "class_type": "VAELoader",
            "inputs": {"vae_name": WAN_VAE},
        },
        "4": {
            "class_type": "LoadVideo",
            "inputs": {"file": video_filename},
        },
        "5": {
            "class_type": "GetVideoComponents",
            "inputs": {"video": ["4", 0]},
        },
        "6": {
            "class_type": "CLIPTextEncode",
            "inputs": {
                "clip": ["2", 0],
                "text": prompt,
            },
        },
        "7": {
            "class_type": "CLIPTextEncode",
            "inputs": {
                "clip": ["2", 0],
                "text": NEGATIVE_PROMPT_VACE,
            },
        },
        "8": {
            "class_type": "WanVaceToVideo",
            "inputs": {
                "positive": ["6", 0],
                "negative": ["7", 0],
                "vae": ["3", 0],
                "control_video": ["5", 0],
                "width": int(width),
                "height": int(height),
                "length": int(frames),
                "batch_size": 1,
                "strength": 1.0,
            },
        },
        "10": {
            "class_type": "ModelSamplingSD3",
            "inputs": {
                "model": ["1", 0],
                "shift": 8,
            },
        },
        "11": {
            "class_type": "KSampler",
            "inputs": {
                "model": ["10", 0],
                "positive": ["8", 0],
                "negative": ["8", 1],
                "latent_image": ["8", 2],
                "seed": int(seed),
                "steps": int(steps),
                "cfg": float(cfg),
                "sampler_name": "uni_pc",
                "scheduler": "simple",
                "denoise": 1.0,
            },
        },
        "12": {
            "class_type": "TrimVideoLatent",
            "inputs": {
                "samples": ["11", 0],
                "trim_amount": ["8", 3],
            },
        },
        "13": {
            "class_type": "VAEDecode",
            "inputs": {
                "samples": ["12", 0],
                "vae": ["3", 0],
            },
        },
        "14": {
            "class_type": "CreateVideo",
            "inputs": {
                "images": ["13", 0],
                "fps": float(FPS_FINAL),
            },
        },
        "15": {
            "class_type": "SaveVideo",
            "inputs": {
                "video": ["14", 0],
                "filename_prefix": "gradio/vace_segments",
                "format": "auto",
                "codec": "auto",
            },
        },
    }

    if reference_filename:
        workflow["9"] = {
            "class_type": "LoadImage",
            "inputs": {"image": reference_filename},
        }
        workflow["8"]["inputs"]["reference_image"] = ["9", 0]

    return workflow


def run_vace(
    video_path,
    face_ref,
    body_ref,
    location_ref,
    prompt,
    width,
    height,
    frames,  # mantido para compatibilidade com o .click existente; cada trecho usa 81
    steps,
    cfg,
    seed,
    progress=gr.Progress(),
):
    if not video_path:
        raise gr.Error("Envie o vídeo de origem.")
    if not prompt or not prompt.strip():
        raise gr.Error("Escreva um prompt para orientar a transformação.")

    width = max(16, (int(width) // 16) * 16)
    height = max(16, (int(height) // 16) * 16)

    if seed is None or int(seed) < 0:
        seed = int.from_bytes(os.urandom(8), "big") % (2**31 - 1)
    else:
        seed = int(seed)

    # Gr.Video pode retornar caminho simples ou objeto/dicionário.
    if isinstance(video_path, dict):
        video_path = video_path.get("path") or video_path.get("name")

    if not video_path or not Path(video_path).is_file():
        raise gr.Error("Não consegui acessar o arquivo do vídeo enviado.")

    pasta_execucao = OUTPUT_DIR / "vace_segments" / uuid.uuid4().hex[:10]
    pasta_execucao.mkdir(parents=True, exist_ok=True)

    manifest_path = pasta_execucao / "manifest.json"
    videos_gerados = []
    prompt_ids = []

    with GPU_LOCK:
        progress(0.01, desc="Preparando o primeiro trecho...")
        print(
            f"[VACE] Início: 6 trechos de até 81 quadros, "
            f"saída a {FPS_FINAL} fps.",
            flush=True,
        )

        # A foto de rosto, corpo e local vira uma referência visual composta.
        reference_board = make_reference_board(
            face_ref,
            body_ref,
            location_ref,
        )
        reference_filename = (
            upload_image_to_comfyui(reference_board)
            if reference_board is not None
            else None
        )

        # O primeiro trecho usa o vídeo enviado. Os seguintes partem do último
        # quadro gerado para tentar continuar a subida da câmera.
        controle_atual = padronizar_video_controle(
            entrada=video_path,
            saida=pasta_execucao / "controle_inicial.mp4",
            largura=width,
            altura=height,
        )

        rotulos = {
            "1": "carregando Wan VACE",
            "2": "carregando encoder de texto",
            "3": "carregando VAE",
            "4": "lendo vídeo de controle",
            "5": "extraindo quadros",
            "6": "codificando prompt",
            "8": "preparando condicionamento VACE",
            "11": "gerando quadros",
            "13": "decodificando quadros",
            "14": "montando trecho de vídeo",
            "15": "salvando trecho",
        }

        for indice in range(NUM_TRECHOS):
            numero = indice + 1
            progress(
                indice / NUM_TRECHOS,
                desc=f"Trecho {numero}/{NUM_TRECHOS}: preparando entrada...",
            )
            print(
                f"[VACE] Preparando trecho {numero}/{NUM_TRECHOS}",
                flush=True,
            )

            # Copia o vídeo de controle para a pasta input do ComfyUI.
            video_filename = copy_video_to_comfyui_input(controle_atual)

            prompt_trecho = prompt_do_trecho(prompt, indice)
            seed_trecho = (seed + indice) % (2**31 - 1)

            workflow = build_vace_workflow(
                video_filename=video_filename,
                reference_filename=reference_filename,
                prompt=prompt_trecho,
                width=width,
                height=height,
                frames=FRAMES_POR_TRECHO,
                steps=int(steps),
                cfg=float(cfg),
                seed=seed_trecho,
            )

            # Mapeia o progresso local de cada workflow para o total da sequência.
            def progresso_trecho(valor, desc=None, i=indice):
                progresso_total = (i + float(valor)) / NUM_TRECHOS
                descricao = desc or "gerando"
                progress(
                    progresso_total,
                    desc=f"Trecho {i + 1}/{NUM_TRECHOS}: {descricao}",
                )

            video_saida, prompt_id = run_comfy_workflow(
                workflow=workflow,
                save_node_id="15",
                media_type="videos",
                output_folder=OUTPUT_DIR / "vace",
                node_labels=rotulos,
                progress=progresso_trecho,
            )

            videos_gerados.append(video_saida)
            prompt_ids.append(prompt_id)

            # Salva o progresso no disco depois de cada trecho.
            manifest_path.write_text(
                json.dumps(
                    {
                        "seed": seed,
                        "fps": FPS_FINAL,
                        "trechos_concluidos": videos_gerados,
                        "prompt_ids": prompt_ids,
                    },
                    ensure_ascii=False,
                    indent=2,
                ),
                encoding="utf-8",
            )

            print(
                f"[VACE] Trecho {numero}/{NUM_TRECHOS} salvo: {video_saida}",
                flush=True,
            )

            # Prepara o último quadro para orientar o próximo segmento.
            if indice < NUM_TRECHOS - 1:
                ultimo_quadro = pasta_execucao / f"ultimo_quadro_{numero}.png"
                extrair_ultimo_quadro(video_saida, ultimo_quadro)

                controle_atual = criar_controle_do_ultimo_quadro(
                    imagem=ultimo_quadro,
                    saida=pasta_execucao / f"controle_{numero + 1}.mp4",
                    largura=width,
                    altura=height,
                )

        progress(0.94, desc="Unindo os seis trechos em um vídeo de 30 segundos...")
        print("[VACE] Unindo e cortando a sequência para 30 segundos.", flush=True)

        video_final = pasta_execucao / "video_final_30s.mp4"
        juntar_trechos_em_30_segundos(
            videos=videos_gerados,
            destino=video_final,
            fps=FPS_FINAL,
        )

        manifest_path.write_text(
            json.dumps(
                {
                    "seed": seed,
                    "fps": FPS_FINAL,
                    "duracao_final_segundos": DURACAO_FINAL_SEGUNDOS,
                    "video_final": str(video_final),
                    "trechos_concluidos": videos_gerados,
                    "prompt_ids": prompt_ids,
                },
                ensure_ascii=False,
                indent=2,
            ),
            encoding="utf-8",
        )

    return (
        str(video_final),
        "Concluído: 6 trechos processados e unidos em um vídeo de 30 segundos "
        f"a {FPS_FINAL} fps. IDs: {', '.join(prompt_ids)}",
    )
```

```python
# Dependência para receber eventos WebSocket do ComfyUI.
# Se necessário, execute antes em outra célula:
# %pip install -q websocket-client

import io
import json
import os
import secrets
import shutil
import time
import uuid
import threading
from pathlib import Path

import gradio as gr
import requests
import websocket
from PIL import Image, ImageOps


# ---------------------------------------------------------------------------
# Configuração
# ---------------------------------------------------------------------------
COMFYUI_URL = os.environ.get("COMFYUI_URL", "http://127.0.0.1:8188").rstrip("/")
COMFYUI_DIR = Path(os.environ.get("COMFYUI_DIR", "ComfyUI_repo"))

OUTPUT_DIR = Path("outputs/gradio")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# Os arquivos precisam existir nas pastas de modelos do ComfyUI.
# Para GPUs com menos VRAM, defina VACE_MODEL com outra variante
# (por exemplo, a versão 1.3B ou uma quantização fp8 que você tenha instalado).
VACE_MODEL = os.environ.get("VACE_MODEL", "wan2.1_vace_14B_fp16.safetensors")
WAN_TEXT_ENCODER = "umt5_xxl_fp8_e4m3fn_scaled.safetensors"
WAN_VAE = "wan_2.1_vae.safetensors"

FLUX_FILL_MODEL = "flux1-fill-dev.safetensors"
FLUX_CLIP_L = "clip_l.safetensors"
FLUX_T5 = "t5xxl_fp8_e4m3fn.safetensors"
FLUX_VAE = "ae.safetensors"

NEGATIVE_PROMPT_VACE = (
    "flicker, jitter, blurry, low quality, distorted, deformed anatomy, "
    "extra limbs, duplicate objects, text, subtitles, watermark"
)

# Lado maior máximo da imagem no Flux Fill (evita estourar a VRAM).
FLUX_MAX_SIDE = int(os.environ.get("FLUX_MAX_SIDE", "1536"))

# Não deixa duas gerações grandes competirem pela VRAM ao mesmo tempo.
GPU_LOCK = threading.Lock()


# ---------------------------------------------------------------------------
# Utilitários de comunicação com o ComfyUI
# ---------------------------------------------------------------------------
def check_comfyui():
    try:
        response = requests.get(f"{COMFYUI_URL}/system_stats", timeout=10)
        response.raise_for_status()
    except requests.RequestException as exc:
        raise gr.Error(
            f"Não consegui acessar o ComfyUI em {COMFYUI_URL}. "
            "Confira se o servidor está ativo."
        ) from exc


def upload_bytes_to_comfyui(data: bytes, filename: str, mime: str) -> str:
    """Envia um arquivo para a pasta input do ComfyUI via HTTP."""
    try:
        response = requests.post(
            f"{COMFYUI_URL}/upload/image",
            files={"image": (filename, data, mime)},
            data={"type": "input", "overwrite": "true"},
            timeout=300,
        )
        response.raise_for_status()
    except requests.RequestException as exc:
        raise gr.Error(f"Falha ao enviar arquivo para o ComfyUI: {exc}") from exc

    return response.json().get("name", filename)


def upload_image_to_comfyui(image: Image.Image) -> str:
    buffer = io.BytesIO()
    image.save(buffer, format="PNG")
    filename = f"gradio_{uuid.uuid4().hex[:12]}.png"
    return upload_bytes_to_comfyui(buffer.getvalue(), filename, "image/png")


def upload_video_to_comfyui(video_path) -> str:
    """
    Envia o vídeo por HTTP (funciona mesmo que o ComfyUI esteja em outra
    máquina). Se falhar e a pasta local do ComfyUI existir, copia o arquivo.
    """
    if isinstance(video_path, dict):
        video_path = video_path.get("path") or video_path.get("name")

    if not video_path or not Path(video_path).is_file():
        raise gr.Error("Não consegui ler o vídeo enviado.")

    source = Path(video_path)
    safe_name = f"gradio_{uuid.uuid4().hex[:12]}{source.suffix.lower()}"
    mime = "video/mp4" if source.suffix.lower() == ".mp4" else "application/octet-stream"

    try:
        return upload_bytes_to_comfyui(source.read_bytes(), safe_name, mime)
    except gr.Error:
        input_dir = COMFYUI_DIR / "input"
        if not input_dir.exists():
            raise
        shutil.copy2(source, input_dir / safe_name)
        return safe_name


def make_reference_board(face_ref, body_ref, location_ref, width, height):
    """
    Combina até três referências lado a lado numa única imagem com a mesma
    proporção do vídeo de saída (sem quadrantes vazios e sem distorção).
    """
    references = [face_ref, body_ref, location_ref]
    references = [img.convert("RGB") for img in references if img is not None]

    if not references:
        return None

    # Uma única referência: usa a imagem direto, sem colagem.
    if len(references) == 1:
        return references[0]

    board = Image.new("RGB", (width, height), (128, 128, 128))
    tile_w = width // len(references)

    for index, image in enumerate(references):
        fitted = ImageOps.contain(image, (tile_w, height))
        px = index * tile_w + (tile_w - fitted.width) // 2
        py = (height - fitted.height) // 2
        board.paste(fitted, (px, py))

    return board


def make_rgba_inpaint_image(image: Image.Image, mask_image: Image.Image) -> Image.Image:
    """
    Interpreta branco na máscara como área a editar e preto como área a preservar.
    Converte a máscara em canal alfa para que o LoadImage do ComfyUI gere a máscara.
    """
    rgb = image.convert("RGB")
    mask = mask_image.convert("L").resize(rgb.size, Image.Resampling.NEAREST)

    # No LoadImage do ComfyUI, a máscara é 1 - alfa.
    alpha = ImageOps.invert(mask)
    rgba = rgb.convert("RGBA")
    rgba.putalpha(alpha)
    return rgba


def fit_for_flux(image: Image.Image, mask: Image.Image):
    """Limita o lado maior e arredonda as dimensões para múltiplos de 16."""
    width, height = image.size
    scale = min(1.0, FLUX_MAX_SIDE / max(width, height))

    new_w = max(16, int(round(width * scale / 16)) * 16)
    new_h = max(16, int(round(height * scale / 16)) * 16)

    if (new_w, new_h) != (width, height):
        image = image.convert("RGB").resize((new_w, new_h), Image.Resampling.LANCZOS)
    mask = mask.convert("L").resize((new_w, new_h), Image.Resampling.NEAREST)
    return image, mask


def get_saved_media(node_output, media_type):
    """
    Localiza o arquivo gravado por um nó SaveVideo/SaveImage.
    O SaveVideo costuma registrar o resultado sob a chave "images"
    (com "animated": [true]), por isso procuramos em várias chaves.
    """
    if not isinstance(node_output, dict):
        return None

    keys = ("videos", "images", "gifs") if media_type == "videos" else (media_type,)

    for container in (node_output, node_output.get("ui", {})):
        if not isinstance(container, dict):
            continue

        for key in keys:
            items = container.get(key, [])
            if isinstance(items, dict):
                items = [items]

            for item in items:
                if isinstance(item, dict) and item.get("filename"):
                    return item

    return None


def get_history_entry(prompt_id):
    response = requests.get(
        f"{COMFYUI_URL}/history/{prompt_id}",
        timeout=30,
    )
    response.raise_for_status()
    return response.json().get(prompt_id)


def extract_execution_error(entry):
    status = entry.get("status", {})
    for message in status.get("messages", []):
        if (
            isinstance(message, (list, tuple))
            and len(message) > 1
            and message[0] == "execution_error"
        ):
            detail = message[1]
            if isinstance(detail, dict):
                return (
                    detail.get("exception_message")
                    or detail.get("message")
                    or json.dumps(detail, ensure_ascii=False)
                )
            return str(detail)
    return None


def run_comfy_workflow(
    workflow,
    save_node_id,
    media_type,
    output_folder,
    node_labels,
    progress,
):
    """
    Envia o workflow e monitora a API WebSocket do ComfyUI.
    Retorna (caminho local do arquivo salvo, prompt_id).
    """
    check_comfyui()

    client_id = str(uuid.uuid4())
    ws_url = (
        COMFYUI_URL.replace("https://", "wss://")
        .replace("http://", "ws://")
        + f"/ws?clientId={client_id}"
    )

    try:
        ws = websocket.create_connection(ws_url, timeout=2)
    except Exception as exc:
        raise gr.Error(
            "Não consegui abrir o WebSocket do ComfyUI. "
            "Confira a URL e a porta do servidor."
        ) from exc

    # A barra de progresso nunca anda para trás.
    last_progress = 0.0

    def advance(value, desc):
        nonlocal last_progress
        last_progress = max(last_progress, value)
        progress(last_progress, desc=desc)

    prompt_id = None
    saved_file = None
    last_history_check = 0
    deadline = time.time() + 3 * 60 * 60

    try:
        response = requests.post(
            f"{COMFYUI_URL}/prompt",
            json={"prompt": workflow, "client_id": client_id},
            timeout=60,
        )

        if not response.ok:
            raise gr.Error(
                "O ComfyUI rejeitou o workflow:\n" + response.text[:5000]
            )

        submitted = response.json()
        if submitted.get("error"):
            details = submitted.get("node_errors", submitted["error"])
            raise gr.Error(
                "O ComfyUI rejeitou o workflow:\n"
                + json.dumps(details, ensure_ascii=False, indent=2)
            )

        prompt_id = submitted.get("prompt_id")
        if not prompt_id:
            raise gr.Error(f"Resposta inesperada do ComfyUI: {submitted}")

        advance(0.08, "Workflow enviado ao ComfyUI.")
        print(f"[ComfyUI] Workflow enviado: {prompt_id}", flush=True)

        while time.time() < deadline:
            try:
                raw_message = ws.recv()
            except Exception:
                raw_message = None

            if isinstance(raw_message, str):
                try:
                    message = json.loads(raw_message)
                except json.JSONDecodeError:
                    message = {}

                event_type = message.get("type")
                data = message.get("data", {})

                if data.get("prompt_id") == prompt_id:
                    if event_type == "executing":
                        node_id = str(data.get("node"))
                        label = node_labels.get(node_id, f"Nó {node_id}")

                        if data.get("node") is None:
                            advance(0.94, "Execução terminada; recuperando o arquivo...")
                            print("[ComfyUI] Nós concluídos.", flush=True)
                        else:
                            advance(0.20, f"Em execução: {label}")
                            print(f"[ComfyUI] Em execução: {label}", flush=True)

                    elif event_type == "progress":
                        value = int(data.get("value", 0))
                        maximum = max(1, int(data.get("max", 1)))
                        percent = min(1.0, value / maximum)
                        advance(
                            0.25 + 0.62 * percent,
                            f"Amostragem: passo {value}/{maximum}",
                        )
                        print(f"[ComfyUI] Sampler: {value}/{maximum}", flush=True)

                    elif event_type == "execution_error":
                        raise gr.Error(
                            "O ComfyUI encontrou um erro durante a execução:\n"
                            + json.dumps(data, ensure_ascii=False, indent=2)
                        )

            # Consulta também o histórico como fallback para versões sem eventos.
            if time.time() - last_history_check >= 3:
                last_history_check = time.time()

                try:
                    entry = get_history_entry(prompt_id)
                except (requests.RequestException, ValueError) as exc:
                    # Falha momentânea de rede: tenta de novo no próximo ciclo.
                    print(f"[ComfyUI] Falha ao consultar histórico: {exc}", flush=True)
                    entry = None

                if entry:
                    error = extract_execution_error(entry)
                    if error:
                        raise gr.Error(f"Erro no ComfyUI:\n{error}")

                    output = entry.get("outputs", {}).get(str(save_node_id), {})
                    saved_file = get_saved_media(output, media_type)

                    if saved_file:
                        break

            time.sleep(0.2)

        if not saved_file:
            raise gr.Error(
                "A execução não retornou um arquivo antes do limite de espera. "
                "Confira a fila e o histórico do ComfyUI."
            )

    finally:
        try:
            ws.close()
        except Exception:
            pass

    advance(0.96, "Baixando o resultado do ComfyUI...")

    try:
        result = requests.get(
            f"{COMFYUI_URL}/view",
            params={
                "filename": saved_file["filename"],
                "subfolder": saved_file.get("subfolder", ""),
                "type": saved_file.get("type", "output"),
            },
            timeout=300,
        )
        result.raise_for_status()
    except requests.RequestException as exc:
        raise gr.Error(f"Falha ao baixar o resultado: {exc}") from exc

    extension = Path(saved_file["filename"]).suffix
    if not extension:
        extension = ".mp4" if media_type in ("videos", "gifs") else ".png"

    output_folder = Path(output_folder)
    output_folder.mkdir(parents=True, exist_ok=True)
    local_path = output_folder / f"{prompt_id}{extension}"
    local_path.write_bytes(result.content)

    advance(1.0, "Concluído.")
    print(f"[ComfyUI] Arquivo salvo em: {local_path}", flush=True)
    return str(local_path), prompt_id


# ---------------------------------------------------------------------------
# Wan VACE (vídeo para vídeo)
# ---------------------------------------------------------------------------
def build_vace_workflow(
    video_filename,
    reference_filename,
    prompt,
    width,
    height,
    frames,
    steps,
    cfg,
    seed,
):
    vace_inputs = {
        "positive": ["6", 0],
        "negative": ["7", 0],
        "vae": ["3", 0],
        "control_video": ["5", 0],
        "width": int(width),
        "height": int(height),
        "length": int(frames),
        "batch_size": 1,
        "strength": 1.0,
    }

    workflow = {
        "1": {
            "class_type": "UNETLoader",
            "inputs": {"unet_name": VACE_MODEL, "weight_dtype": "default"},
        },
        "2": {
            "class_type": "CLIPLoader",
            "inputs": {
                "clip_name": WAN_TEXT_ENCODER,
                "type": "wan",
                "device": "default",
            },
        },
        "3": {
            "class_type": "VAELoader",
            "inputs": {"vae_name": WAN_VAE},
        },
        "4": {
            "class_type": "LoadVideo",
            "inputs": {"file": video_filename},
        },
        "5": {
            "class_type": "GetVideoComponents",
            "inputs": {"video": ["4", 0]},
        },
        "6": {
            "class_type": "CLIPTextEncode",
            "inputs": {"clip": ["2", 0], "text": prompt},
        },
        "7": {
            "class_type": "CLIPTextEncode",
            "inputs": {"clip": ["2", 0], "text": NEGATIVE_PROMPT_VACE},
        },
        "10": {
            "class_type": "ModelSamplingSD3",
            "inputs": {"model": ["1", 0], "shift": 8},
        },
        "11": {
            "class_type": "KSampler",
            "inputs": {
                "model": ["10", 0],
                "positive": ["8", 0],
                "negative": ["8", 1],
                "latent_image": ["8", 2],
                "seed": int(seed),
                "steps": int(steps),
                "cfg": float(cfg),
                "sampler_name": "uni_pc",
                "scheduler": "simple",
                "denoise": 1.0,
            },
        },
        "12": {
            "class_type": "TrimVideoLatent",
            "inputs": {"samples": ["11", 0], "trim_amount": ["8", 3]},
        },
        "13": {
            "class_type": "VAEDecode",
            "inputs": {"samples": ["12", 0], "vae": ["3", 0]},
        },
        "14": {
            "class_type": "CreateVideo",
            "inputs": {"images": ["13", 0], "fps": ["5", 2], "codec": "none"},
        },
        "15": {
            "class_type": "SaveVideo",
            "inputs": {
                "video": ["14", 0],
                "filename_prefix": "gradio/vace_v2v",
                "format": "auto",
                "codec": "auto",
            },
        },
    }

    # A referência só entra no grafo se existir (evita enviar null ao ComfyUI).
    if reference_filename:
        workflow["9"] = {
            "class_type": "LoadImage",
            "inputs": {"image": reference_filename},
        }
        vace_inputs["reference_image"] = ["9", 0]

    workflow["8"] = {"class_type": "WanVaceToVideo", "inputs": vace_inputs}
    return workflow


def run_vace(
    video_path,
    face_ref,
    body_ref,
    location_ref,
    prompt,
    width,
    height,
    frames,
    steps,
    cfg,
    seed,
    progress=gr.Progress(),
):
    if not video_path:
        raise gr.Error("Envie o vídeo de origem.")
    if not prompt or not prompt.strip():
        raise gr.Error("Escreva um prompt para orientar a transformação.")

    width = max(16, (int(width) // 16) * 16)
    height = max(16, (int(height) // 16) * 16)
    frames = max(5, 4 * ((int(frames) - 1) // 4) + 1)

    if seed is None or int(seed) < 0:
        seed = int.from_bytes(os.urandom(8), "big") % (2**63 - 1)

    with GPU_LOCK:
        progress(0.02, desc="Preparando o vídeo e as referências...")
        video_filename = upload_video_to_comfyui(video_path)

        reference_board = make_reference_board(
            face_ref, body_ref, location_ref, width, height
        )

        reference_filename = None
        if reference_board is not None:
            reference_filename = upload_image_to_comfyui(reference_board)

        workflow = build_vace_workflow(
            video_filename=video_filename,
            reference_filename=reference_filename,
            prompt=prompt.strip(),
            width=width,
            height=height,
            frames=frames,
            steps=int(steps),
            cfg=float(cfg),
            seed=int(seed),
        )

        node_labels = {
            "1": "carregando o modelo Wan VACE",
            "2": "carregando o encoder de texto",
            "3": "carregando o VAE",
            "4": "lendo o vídeo",
            "5": "extraindo quadros do vídeo",
            "6": "codificando o prompt",
            "8": "preparando condicionamento VACE",
            "11": "gerando os quadros",
            "13": "decodificando os quadros",
            "14": "montando o vídeo",
            "15": "salvando o vídeo",
        }

        path, prompt_id = run_comfy_workflow(
            workflow=workflow,
            save_node_id="15",
            media_type="videos",
            output_folder=OUTPUT_DIR / "vace",
            node_labels=node_labels,
            progress=progress,
        )

    return path, f"VACE concluído. ID: {prompt_id}"


# ---------------------------------------------------------------------------
# Flux Fill (inpainting)
# ---------------------------------------------------------------------------
def build_flux_fill_workflow(image_filename, prompt, steps, seed):
    return {
        "1": {
            "class_type": "UNETLoader",
            "inputs": {"unet_name": FLUX_FILL_MODEL, "weight_dtype": "default"},
        },
        "2": {
            "class_type": "DualCLIPLoader",
            "inputs": {
                "clip_name1": FLUX_CLIP_L,
                "clip_name2": FLUX_T5,
                "type": "flux",
                "device": "default",
            },
        },
        "3": {
            "class_type": "VAELoader",
            "inputs": {"vae_name": FLUX_VAE},
        },
        "4": {
            "class_type": "LoadImage",
            "inputs": {"image": image_filename},
        },
        "5": {
            "class_type": "CLIPTextEncode",
            "inputs": {"clip": ["2", 0], "text": prompt},
        },
        "6": {
            "class_type": "FluxGuidance",
            "inputs": {"conditioning": ["5", 0], "guidance": 30.0},
        },
        "7": {
            "class_type": "ConditioningZeroOut",
            "inputs": {"conditioning": ["5", 0]},
        },
        "8": {
            "class_type": "DifferentialDiffusion",
            "inputs": {"model": ["1", 0]},
        },
        "9": {
            "class_type": "InpaintModelConditioning",
            "inputs": {
                "positive": ["6", 0],
                "negative": ["7", 0],
                "vae": ["3", 0],
                "pixels": ["4", 0],
                "mask": ["4", 1],
                "noise_mask": True,
            },
        },
        "10": {
            "class_type": "KSampler",
            "inputs": {
                "model": ["8", 0],
                "positive": ["9", 0],
                "negative": ["9", 1],
                "latent_image": ["9", 2],
                "seed": int(seed),
                "steps": int(steps),
                "cfg": 1.0,
                "sampler_name": "euler",
                "scheduler": "simple",
                "denoise": 1.0,
            },
        },
        "11": {
            "class_type": "VAEDecode",
            "inputs": {"samples": ["10", 0], "vae": ["3", 0]},
        },
        "12": {
            "class_type": "SaveImage",
            "inputs": {"images": ["11", 0], "filename_prefix": "gradio/flux_fill"},
        },
    }


def run_flux_fill(
    image,
    mask,
    prompt,
    steps,
    seed,
    progress=gr.Progress(),
):
    if image is None:
        raise gr.Error("Envie a imagem que deseja editar.")
    if mask is None:
        raise gr.Error("Envie uma máscara em preto e branco.")
    if not prompt or not prompt.strip():
        raise gr.Error("Escreva o que deve aparecer na área mascarada.")

    if seed is None or int(seed) < 0:
        seed = int.from_bytes(os.urandom(8), "big") % (2**63 - 1)

    with GPU_LOCK:
        progress(0.02, desc="Preparando imagem e máscara...")
        image, mask = fit_for_flux(image, mask)
        rgba_image = make_rgba_inpaint_image(image, mask)
        image_filename = upload_image_to_comfyui(rgba_image)

        workflow = build_flux_fill_workflow(
            image_filename=image_filename,
            prompt=prompt.strip(),
            steps=int(steps),
            seed=int(seed),
        )

        node_labels = {
            "1": "carregando Flux Fill",
            "2": "carregando encoders de texto",
            "3": "carregando VAE",
            "4": "lendo imagem e máscara",
            "5": "codificando o prompt",
            "9": "preparando inpainting",
            "10": "gerando a imagem",
            "11": "decodificando imagem",
            "12": "salvando resultado",
        }

        path, prompt_id = run_comfy_workflow(
            workflow=workflow,
            save_node_id="12",
            media_type="images",
            output_folder=OUTPUT_DIR / "flux_fill",
            node_labels=node_labels,
            progress=progress,
        )

    return path, path, f"Flux Fill concluído. ID: {prompt_id}"


# ---------------------------------------------------------------------------
# Interface
# ---------------------------------------------------------------------------
with gr.Blocks(
    theme=gr.themes.Soft(),
    title="Wan VACE e Flux Fill",
) as demo:
    gr.Markdown(
        """
        # Painel de vídeo e edição de imagem

        O V2V usa Wan VACE para transformar um vídeo com orientação textual e
        referências visuais. O Flux Fill edita uma imagem usando uma máscara.
        O progresso do ComfyUI é exibido durante a execução.
        """
    )

    with gr.Tabs():
        with gr.Tab("Vídeo para vídeo — Wan/VACE"):
            gr.Markdown(
                "O VACE recebe um vídeo de controle e até três referências. "
                "Com mais de uma referência, elas são dispostas lado a lado "
                "numa única imagem com a proporção do vídeo de saída."
            )

            with gr.Row():
                with gr.Column():
                    v2v_video = gr.Video(
                        label="Vídeo de origem",
                        format="mp4",
                        sources=["upload"],
                    )

                    v2v_prompt = gr.Textbox(
                        label="Prompt para a transformação",
                        placeholder=(
                            "Descreva a aparência desejada e o que deve mudar. "
                            "Ex.: manter o movimento da câmera e substituir "
                            "o cenário por uma floresta ao entardecer."
                        ),
                        lines=4,
                    )

                    with gr.Row():
                        v2v_face = gr.Image(
                            label="Referência de rosto (opcional)",
                            type="pil",
                        )
                        v2v_body = gr.Image(
                            label="Referência de corpo/roupa (opcional)",
                            type="pil",
                        )
                        v2v_location = gr.Image(
                            label="Referência de local/estilo (opcional)",
                            type="pil",
                        )

                    with gr.Row():
                        v2v_width = gr.Slider(
                            512, 1280, value=832, step=16,
                            label="Largura de saída",
                        )
                        v2v_height = gr.Slider(
                            384, 720, value=480, step=16,
                            label="Altura de saída",
                        )

                    v2v_frames = gr.Dropdown(
                        choices=[33, 49, 81],
                        value=49,
                        label="Frames (4n+1)",
                    )

                    with gr.Row():
                        v2v_steps = gr.Slider(
                            10, 40, value=20, step=1,
                            label="Passos",
                        )
                        v2v_cfg = gr.Slider(
                            1, 10, value=6, step=0.5,
                            label="CFG",
                        )

                    v2v_seed = gr.Number(
                        value=-1,
                        precision=0,
                        label="Seed (-1 para aleatória)",
                    )
                    v2v_button = gr.Button(
                        "Gerar vídeo V2V",
                        variant="primary",
                    )

                with gr.Column():
                    v2v_output = gr.Video(label="Vídeo gerado")
                    v2v_status = gr.Textbox(label="Status", lines=2)

            v2v_button.click(
                fn=run_vace,
                inputs=[
                    v2v_video,
                    v2v_face,
                    v2v_body,
                    v2v_location,
                    v2v_prompt,
                    v2v_width,
                    v2v_height,
                    v2v_frames,
                    v2v_steps,
                    v2v_cfg,
                    v2v_seed,
                ],
                outputs=[v2v_output, v2v_status],
            )

        with gr.Tab("Inpainting — Flux Fill"):
            gr.Markdown(
                "Envie uma imagem e uma máscara. **Branco = área a editar; "
                "preto = preservar.** Imagens grandes são reduzidas "
                f"automaticamente (lado maior máx. {FLUX_MAX_SIDE}px). "
                "O resultado pode ser baixado e enviado depois como "
                "referência na aba V2V."
            )

            with gr.Row():
                with gr.Column():
                    flux_image = gr.Image(
                        label="Imagem original",
                        type="pil",
                    )
                    flux_mask = gr.Image(
                        label="Máscara em preto e branco",
                        type="pil",
                    )
                    flux_prompt = gr.Textbox(
                        label="O que deve aparecer na área branca?",
                        placeholder=(
                            "Ex.: preencher a área com uma jaqueta vermelha "
                            "de couro, mantendo iluminação e perspectiva."
                        ),
                        lines=3,
                    )
                    flux_steps = gr.Slider(
                        10, 40, value=20, step=1,
                        label="Passos",
                    )
                    flux_seed = gr.Number(
                        value=-1,
                        precision=0,
                        label="Seed (-1 para aleatória)",
                    )
                    flux_button = gr.Button(
                        "Editar imagem com Flux Fill",
                        variant="primary",
                    )

                with gr.Column():
                    flux_output = gr.Image(label="Imagem editada")
                    flux_download = gr.File(label="Baixar imagem editada")
                    flux_status = gr.Textbox(label="Status", lines=2)

            flux_button.click(
                fn=run_flux_fill,
                inputs=[
                    flux_image,
                    flux_mask,
                    flux_prompt,
                    flux_steps,
                    flux_seed,
                ],
                outputs=[
                    flux_output,
                    flux_download,
                    flux_status,
                ],
            )


# ---------------------------------------------------------------------------
# Lançamento com autenticação
# ---------------------------------------------------------------------------
# Defina GRADIO_USER e GRADIO_PASS no ambiente para usar suas credenciais.
# Se não definir, uma senha aleatória é gerada e impressa abaixo.
auth_user = os.environ.get("GRADIO_USER", "admin")
auth_pass = os.environ.get("GRADIO_PASS") or secrets.token_urlsafe(10)

print(f"[Gradio] Login: {auth_user} | Senha: {auth_pass}", flush=True)

# Em Colab, share=True cria um link público temporário (protegido por login).
demo.queue().launch(
    share=True,
    debug=False,
    auth=(auth_user, auth_pass),
)
```
